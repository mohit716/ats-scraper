```python
import json
import scrapy

class SmartRecruitersBase(scrapy.Spider):
    company_identifier = ""  # e.g., "smartrecruiters", "visa", "plaid"

    def start_requests(self):
        url = f"https://api.smartrecruiters.com/v1/companies/{self.company_identifier}/postings?limit=100&offset=0&destination=PUBLIC"
        yield scrapy.Request(url=url, callback=self.parse_postings)

    def parse_postings(self, response):
        data = json.loads(response.text)
        postings = data.get("content", [])

        for item in postings:
            # Construct the public job URL using the company identifier and posting ID
            posting_id = item.get("id")
            job_href = f"https://jobs.smartrecruiters.com/{self.company_identifier}/{posting_id}" if posting_id else ""

            # Standardize location output
            location = item.get("location", {})
            job_city_des = location.get("fullLocation") or ", ".join(
                filter(None, [location.get("city"), location.get("region"), location.get("country")])
            )

            yield {
                "internalType": "",
                "category_name": item.get("typeOfEmployment", {}).get("label", ""),
                "company_name": item.get("company", {}).get("name", self.company_name),
                "job_title": item.get("name", ""),
                "job_href": job_href,
                "job_city_des": job_city_des,
                "details_job": item.get("department", {}).get("label", "")
            }
```

```python
import requests
import pandas as pd

company_identifier = "smartrecruiters"
url = f"https://api.smartrecruiters.com/v1/companies/{company_identifier}/postings?limit=100&offset=0&destination=PUBLIC"

# Fetch data
response = requests.get(url)
data = response.json()

jobs = []
for item in data.get("content", []):
    posting_id = item.get("id")
    job_href = f"https://jobs.smartrecruiters.com/{company_identifier}/{posting_id}" if posting_id else ""
    
    location = item.get("location", {})
    job_city_des = location.get("fullLocation") or ", ".join(
        filter(None, [location.get("city"), location.get("region"), location.get("country")])
    )
    
    jobs.append({
        "internalType": "",
        "category_name": item.get("typeOfEmployment", {}).get("label", ""),
        "company_name": item.get("company", {}).get("name", company_identifier),
        "job_title": item.get("name", ""),
        "job_href": job_href,
        "job_city_des": job_city_des,
        "details_job": item.get("department", {}).get("label", "")
    })

# Convert to DataFrame to view nicely in Jupyter
df = pd.DataFrame(jobs)
df
```
